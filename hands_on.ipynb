```python
import pandas as pd

# Carregando o DataFrame
df = pd.read_csv('pet_adoption_data.csv')

# Dicionários de tradução para colunas e valores categóricos
col_mapping = {
    'PetID': 'ID_Pet',
    'PetType': 'Tipo_Pet',
    'Breed': 'Raca',
    'AgeMonths': 'Idade_Meses',
    'Color': 'Cor',
    'Size': 'Porte',
    'WeightKg': 'Peso_Kg',
    'Vaccinated': 'Vacinado',
    'HealthCondition': 'Condicao_Saude',
    'TimeInShelterDays': 'Dias_No_Abrigo',
    'AdoptionFee': 'Taxa_Adocao',
    'PreviousOwner': 'Dono_Anterior',
    'AdoptionLikelihood': 'Probabilidade_Adocao'
}

type_mapping = {'Bird': 'Passaro', 'Rabbit': 'Coelho', 'Dog': 'Cachorro', 'Cat': 'Gato'}
breed_mapping = {'Parakeet': 'Periquito', 'Rabbit': 'Coelho', 'Golden Retriever': 'Golden Retriever', 'Labrador': 'Labrador', 'Siamese': 'Siames', 'Persian': 'Persa', 'Poodle': 'Poodle'}
color_mapping = {'Orange': 'Laranja', 'White': 'Branco', 'Gray': 'Cinza', 'Brown': 'Marrom', 'Black': 'Preto'}
size_mapping = {'Large': 'Grande', 'Medium': 'Medio', 'Small': 'Pequeno'}

# Traduzindo as colunas usando rename com inplace=True (forma recomendada)
df.rename(columns=col_mapping, inplace=True)

# Traduzindo as linhas/valores das colunas categóricas sem inplace para evitar avisos (FutureWarning)
df['Tipo_Pet'] = df['Tipo_Pet'].replace(type_mapping)
df['Raca'] = df['Raca'].replace(breed_mapping)
df['Cor'] = df['Cor'].replace(color_mapping)
df['Porte'] = df['Porte'].replace(size_mapping)
```
